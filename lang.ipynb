```python
from pydantic import BaseModel, field_validator

class Instrument(BaseModel):
    tag: str
    current_ma: float
    @field_validator("tag")
    @classmethod
    def clean_tag(cls, value):
        return value.strip().upper()

    @field_validator("current_ma")
    @classmethod
    def check_current(cls, value):
        if value < 4 or value > 20:
            raise ValueError("current_ma must be between 4 and 20 mA")
        return value
```

```python
temperature = Instrument(  tag = '561bf1', current_ma = 20 )
```

```python
from pydantic import BaseModel, field_validator


class Address(BaseModel):
    street: str
    city: str
    country: str
    postal_code: str

    @field_validator("street", "city", "country")
    @classmethod
    def must_not_be_empty(cls, value):
        value = value.strip()

        if not value:
            raise ValueError("Field cannot be empty")

        return value

    @field_validator("postal_code")
    @classmethod
    def validate_postal_code(cls, value):
        value = value.strip()

        if len(value) < 4:
            raise ValueError(
                "Postal code must contain at least 4 characters"
            )

        return value
```

```python
address = Address(
    street="King Fahd Road",
    city="Riyadh",
    country="Saudi Arabia",
    postal_code="12345"
)

print(address)
```

### Splitting by Recursive Character Text Splitter

```python
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=1000,
    chunk_overlap=200
)

chunks = text_splitter.split_documents(documents)

print("Number of chunks:", len(chunks))
```

```python
for i,doc in enumerate(chunks[5:10]):
    print(f"Chunk number {i+5}")
    print(doc.page_content)

```

```python
embeddings = OllamaEmbeddings(
    model="nomic-embed-text"
)
cleaned_chunks = filter_complex_metadata(chunks)
```

```python
vector_store = Chroma.from_documents(
    documents=cleaned_chunks,
    embedding=embeddings,
    persist_directory="./instrumentation_chroma_db",
    collection_name="instrumentation"
) 
```

```python
retriever = vector_store.as_retriever(
    search_kwargs = {"k":4}
)
```

```python
llm = ChatOllama(
    model="qwen3.5:9b",
    temperature=0,
    num_ctx=8192
)
```

```python
prompt = ChatPromptTemplate.from_messages([
    (
        "system",
        """
        You are an experienced industrial instrumentation instructor.

        Answer the user's question using only the provided context.

        Explain the answer clearly in engineering language.

        If the answer cannot be found in the context, say:
        "The supplied document does not provide enough information."

        Context:
        {context}
        """
    ),
    (
        "human",
        "{question}"
    )
])
```

```python
from langchain_core.runnables import RunnablePassthrough
from langchain_core.output_parsers import StrOutputParser
def ask_rag(question):

    # Retrieve relevant chunks
    docs = retriever.invoke(question)

    # Combine retrieved text
    context = "\n\n".join(
        doc.page_content for doc in docs
    )

    # Create prompt
    messages = prompt.invoke({
        "context": context,
        "question": question
    })

    # Send to Qwen
    response = llm.invoke(messages)

    return response.content
```

Check schenck controller

```python
answer = ask_rag(
    "what is e2 error "
)

print(answer)
```
