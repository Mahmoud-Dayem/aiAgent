```python
import torch
from transformers import (
    AutoTokenizer,
    AutoModelForSequenceClassification,
)

checkpoint = (
    "distilbert-base-uncased-finetuned-sst-2-english"
)

# Load matching tokenizer and trained classification model
tokenizer = AutoTokenizer.from_pretrained(checkpoint)

model = AutoModelForSequenceClassification.from_pretrained(
    checkpoint
)

# Select device
device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

model = model.to(device)
model.eval()

texts = [
    "I enjoyed learning this lesson.",
    "This was a terrible experience.",
]

# Convert text into rectangular PyTorch tensors
inputs = tokenizer(
    texts,
    padding=True,
    truncation=True,
    return_tensors="pt",
)

# Move all tensors to the same device as the model
inputs = {
    name: tensor.to(device)
    for name, tensor in inputs.items()
}

# Run inference
with torch.no_grad():
    outputs = model(**inputs)

# Raw model scores
logits = outputs.logits

# Convert scores to probabilities
probabilities = torch.softmax(logits, dim=-1)

# Get class index with highest probability
predicted_ids = torch.argmax(
    probabilities,
    dim=-1
)

for text, predicted_id, probability_row in zip(
    texts,
    predicted_ids,
    probabilities
):
    class_id = predicted_id.item()
    label = model.config.id2label[class_id]
    confidence = probability_row[class_id].item()

    print(f"Text: {text}")
    print(f"Label: {label}")
    print(f"Confidence: {confidence:.2%}")
    print()
```

```python
import torch
import evaluate

from datasets import load_dataset
from torch.optim import AdamW
from torch.utils.data import DataLoader
from tqdm.auto import tqdm

from transformers import (
    AutoTokenizer,
    AutoModelForSequenceClassification,
    DataCollatorWithPadding,
    get_scheduler,
)

# 1. Load dataset
raw_datasets = load_dataset(
    "nyu-mll/glue",
    "mrpc"
)

# 2. Load tokenizer
checkpoint = "bert-base-uncased"

tokenizer = AutoTokenizer.from_pretrained(
    checkpoint
)

# 3. Tokenization function
def tokenize_function(batch):
    return tokenizer(
        batch["sentence1"],
        batch["sentence2"],
        truncation=True
    )

# 4. Tokenize all splits
tokenized_datasets = raw_datasets.map(
    tokenize_function,
    batched=True
)

# 5. Remove unused columns
tokenized_datasets = tokenized_datasets.remove_columns(
    ["sentence1", "sentence2", "idx"]
)

# 6. Rename label
tokenized_datasets = tokenized_datasets.rename_column(
    "label",
    "labels"
)

# 7. Return PyTorch tensors
tokenized_datasets.set_format("torch")

# 8. Dynamic padding
data_collator = DataCollatorWithPadding(
    tokenizer=tokenizer
)

# 9. DataLoaders
train_dataloader = DataLoader(
    tokenized_datasets["train"],
    shuffle=True,
    batch_size=8,
    collate_fn=data_collator,
)

eval_dataloader = DataLoader(
    tokenized_datasets["validation"],
    batch_size=8,
    collate_fn=data_collator,
)

# 10. Model
model = AutoModelForSequenceClassification.from_pretrained(
    checkpoint,
    num_labels=2
)

# 11. Device
device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

model.to(device)

# 12. Optimizer
optimizer = AdamW(
    model.parameters(),
    lr=5e-5
)

# 13. Scheduler
num_epochs = 3

num_training_steps = (
    num_epochs * len(train_dataloader)
)

lr_scheduler = get_scheduler(
    "linear",
    optimizer=optimizer,
    num_warmup_steps=0,
    num_training_steps=num_training_steps,
)

# 14. Training
progress_bar = tqdm(
    range(num_training_steps)
)

model.train()

for epoch in range(num_epochs):
    for batch in train_dataloader:
        batch = {
            key: value.to(device)
            for key, value in batch.items()
        }

        outputs = model(**batch)
        loss = outputs.loss

        loss.backward()

        torch.nn.utils.clip_grad_norm_(
            model.parameters(),
            max_norm=1.0
        )

        optimizer.step()
        lr_scheduler.step()
        optimizer.zero_grad()

        progress_bar.update(1)

# 15. Evaluation
metric = evaluate.load(
    "glue",
    "mrpc"
)

model.eval()

for batch in eval_dataloader:
    batch = {
        key: value.to(device)
        for key, value in batch.items()
    }

    with torch.no_grad():
        outputs = model(**batch)

    predictions = torch.argmax(
        outputs.logits,
        dim=-1
    )

    metric.add_batch(
        predictions=predictions,
        references=batch["labels"]
    )

result = metric.compute()
print(result)
```

```python
from transformers import Trainer, TrainingArguments

training_args = TrainingArguments(
    output_dir="./results",
    learning_rate=5e-5,
    per_device_train_batch_size=8,
    per_device_eval_batch_size=8,
    num_train_epochs=3,
    eval_strategy="epoch",
    save_strategy="epoch",
)

trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=tokenized_datasets["train"],
    eval_dataset=tokenized_datasets["validation"],
    data_collator=data_collator,
    processing_class=tokenizer,
)

trainer.train()
```

1. Tiny model

Suppose the model is:

y
^
	​

=w×x

Where:

x = input
w = model weight
y
^
	​

 = prediction

Let:

Input x = 2
Correct answer y = 10
Initial weight w = 3

The model predicts:

y^ =3×2=6 
 
	​
Prediction = 6
Correct value = 10

2. Calculate the loss

For this regression example, we will use squared error:

L=(y^- y) **2
 

Substitute the values:

L=(6−10) **2 
 
L=(−4)
2
loss = 16

So the loss is: 16
The model now knows how wrong it is, but it does not yet know how to change the weight.

That is the job of backward propagation.


```python
import torch

# One trainable weight
w = torch.tensor(3.0, requires_grad=True)

# Input and correct answer
x = torch.tensor(2.0)
y = torch.tensor(10.0)

# Forward pass
prediction = w * x

# Loss
loss = (prediction - y) ** 2

print("Prediction:", prediction.item())
print("Loss:", loss.item())
```

```python
loss.backward()

print("Gradient:", w.grad.item())
```

```python
learning_rate = 0.1

with torch.no_grad():
    w -= learning_rate * w.grad

print("Updated weight:", w.item())
```

```python
import torch

w = torch.tensor(3.0, requires_grad=True)

x = torch.tensor(2.0)
y = torch.tensor(10.0)

learning_rate = 0.1

# Forward pass
prediction = w * x

# Calculate loss
loss = (prediction - y) ** 2

# Backward propagation
loss.backward()

print("Before update")
print("Weight:", w.item())
print("Prediction:", prediction.item())
print("Loss:", loss.item())
print("Gradient:", w.grad.item())

# Optimizer-like update
with torch.no_grad():
    w -= learning_rate * w.grad

# Clear gradient
w.grad.zero_()

print("\nAfter update")
print("New weight:", w.item())
print("New prediction:", (w * x).item())
print("New loss:", ((w * x - y) ** 2).item())
```
