```python
import datetime
import pytz
from smolagents import  DuckDuckGoSearchTool, InferenceClientModel

from smolagents import (
    CodeAgent,
    TransformersModel,b
    tool,
)


@tool
def get_current_time_in_timezone(timezone: str) -> str:
    """
    Gets the current date and time in a specified timezone.

    Args:
        timezone: Valid timezone name, such as Asia/Riyadh,
            Europe/London, or America/New_York.

    Returns:
        The current local date and time in the requested timezone.
    """
    try:
        tz = pytz.timezone(timezone)
        local_time = datetime.datetime.now(tz)

        return (
            f"The current time in {timezone} is "
            f"{local_time.strftime('%Y-%m-%d %H:%M:%S')}."
        )

    except pytz.UnknownTimeZoneError:
        return (
            f"Unknown timezone: {timezone}. "
            "Use a valid timezone such as Asia/Riyadh."
        )

    except Exception as error:
        return f"An error occurred: {error}"


model = TransformersModel(
    model_id="HuggingFaceTB/SmolLM2-1.7B-Instruct",b
    device_map="auto",
    torch_dtype="auto",
    max_new_tokens=512,
    temperature=0.2,
)
web_search_tool = DuckDuckGoSearchTool(
    max_results=5,
    rate_limit=1.0,
)



agent = CodeAgent(
    model=model,
    tools=[get_current_time_in_timezone, web_search_tool],
    max_steps=5,
    verbosity_level=2,
)


# result = agent.run(
#     "What is the current time in Asia/Riyadh?"
# )
result =agent.run("Search for the best music recommendations for a party at the Wayne's mansion.")

print("\nFinal answer:")
print(result)
```

```python
import os

from smolagents import (
    ToolCallingAgent,
    WebSearchTool,
    InferenceClientModel,
    DuckDuckGoSearchTool
)

# Set this once in Windows or use huggingface-cli login
# os.environ["HF_TOKEN"] = "hf_your_token_here"

model = InferenceClientModel(
    model_id="Qwen/Qwen2.5-Coder-32B-Instruct",
    max_tokens=1000,
    temperature=0.2,
)

agent = ToolCallingAgent(
    tools=[WebSearchTool(max_results=5)],
    model=model,
    max_steps=5,
    verbosity_level=2,
)

result = agent.run(
    "Search the web and recommend suitable music for a party "
    "at Wayne's mansion."
)

print(result)
```

```python
def read_file(file_path: str) -> str:
    """Read the contents of a file at the specified path."""
    with open(file_path, 'r') as f:
        return f.read()
```
