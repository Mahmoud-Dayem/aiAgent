```python
import faiss
import numpy as np

vectors = np.array([
    [1.0, 2.0],
    [2.0, 3.0],
    [10.0, 20.0]
], dtype="float32")

index = faiss.IndexFlatL2(2)

index.add(vectors)

query = np.array([
    [1.6, 3.6]
], dtype="float32")

distances, indices = index.search(query, k=2)

print(indices)
print(distances)
```
